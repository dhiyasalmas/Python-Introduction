# Soal 5 (5 menit)

Buatkan program untuk menyelesaikan persamaan
$$ \int^{1}_{0} \frac{4}{1+x^{2}}$$

Dengan menggunakan metode trapezoidal,

$$ \int_{a}^{b} f(x) dx = \frac{h}{2} [f(a)+2 \sum_{i=1}^{n-1} f(a+ih) +f(b)] $$

dimana

$$ h = \frac{b-a}{n}$$

# Soal 6 (5 menit)

Buatkan program untuk menyelesaikan persamaan
$$ \int^{1}_{0} \frac{4}{1+x^{2}}$$

Dengan menggunakan metode midpoint,

$$ \int_{a}^{b} f(x) dx = h \sum_{i=0}^{n-1} f(a+(i+\frac{1}{2})h) $$

dimana

$$ h = \frac{b-a}{n}$$

# Soal 7 (5 menit)

Buatkan program untuk menyelesaikan persamaan
$$ \int^{1}_{0} \frac{4}{1+x^{2}}$$

Dengan menggunakan metode simpson 1/3,

$$ \int_{a}^{b} f(x) dx = \frac{h}{3} [f(a)+ 4 \sum_{i=1,3,5,...}^{n-1} f(a+ih) + 2 \sum_{i=2,4,6,...}^{n-2} f(a+ih) +f(b)] $$

dimana

$$ h = \frac{b-a}{n}$$
