**Spektrum Energi Neutron**

$$ N = \int_{E_1}^{E_2} \phi(E) dE  $$

dengan $ \phi (E) = C E e^{\frac{E}{kT}}$

- E: energi neutron (eV)
- kT: energi rata-rata (misalnya 0.0253 eV untuk suhu 300 K)
- C: konstanta normalisasi (diabaikan)

dengan
- kT : 0.0253 eV (neutron termal pada 300 K)
- E : energi dari 0 s.d 0.3 eV dengan 100 titik
